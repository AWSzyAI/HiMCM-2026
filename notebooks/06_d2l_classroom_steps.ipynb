## 教学简化：线性回归准备
先猜结果，再运行；两人各解释一句。

```python
from torch import nn
x = torch.linspace(0, 1, 20).reshape(-1, 1)
y = 2 * x + 1
torch.manual_seed(0)
model = nn.Linear(1, 1)
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
```

## 教学简化：训练循环
先猜结果，再运行；两人各解释一句。

```python
losses = []
for epoch in range(400):
    optimizer.zero_grad()
    loss = loss_fn(model(x), y)
    loss.backward()
    optimizer.step()
    losses.append(loss.item())
print(model.weight.item(), model.bias.item())
```

## 图：实际训练结果
先猜结果，再运行；两人各解释一句。

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(11, 3.7))
axes[0].plot(losses, color='#117e80', linewidth=2.5)
axes[0].set(xlabel='Epoch', ylabel='MSE loss', title='Actual training loss')
axes[0].set_yscale('log')
axes[1].scatter(x.detach().numpy(), y.detach().numpy(), label='Teaching data', color='#5374a5')
with torch.no_grad():
    prediction = model(x)
axes[1].plot(x.detach().numpy(), prediction.numpy(), color='#117e80', label='Trained prediction', linewidth=2.5)
axes[1].set(xlabel='x', ylabel='y', title='Known rule: y = 2x + 1')
axes[1].legend(frameon=False)
for ax in axes:
    ax.spines[['top', 'right']].set_visible(False)
    ax.grid(alpha=0.12)
fig.tight_layout()
plt.show()
```
