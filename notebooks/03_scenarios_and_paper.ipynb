```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

BASE = Path.cwd()
if BASE.name == 'notebooks':
    BASE = BASE.parent
TABLES = BASE.parent / '2026-09-17 第一次课' / '重写论文' / 'analysis' / 'tables'
summary = BASE.parent / '2026-09-17 第一次课' / '重写论文' / 'analysis' / 'SUMMARY.md'
print(summary.read_text(encoding='utf-8'))
```

```python
scenarios = pd.read_csv(TABLES / 'fig05_2030.csv')
scenarios.head()
```

```python
plt.figure(figsize=(7, 4.5))
for name, group in scenarios.groupby('scenario', sort=False):
    plt.plot(group['year'], group['C_tCO2e'] / 1e6, marker='o', label=name)
plt.xlabel('Year')
plt.ylabel('Conditional emissions (Mt CO2e)')
plt.title('2030 scenario comparison')
plt.legend()
plt.show()
```

```python
final_2030 = scenarios[scenarios['year'] == 2030][['scenario', 'g', 'e', 'd', 'C_tCO2e']].copy()
final_2030['C_MtCO2e'] = final_2030['C_tCO2e'] / 1e6
final_2030
```

```python
matched = pd.read_csv(TABLES / 'matched_systems_2024stock_2023factors.csv')
energy_twh = matched['E_facility_GWh'].sum() / 1000
carbon_mt = matched['C_tCO2e'].sum() / 1e6
n = len(matched)

paragraph = f"""
For the {n} TOP500 systems with reported power and matched 2023 national grid factors,
the conditional annual facility energy is {energy_twh:.3f} TWh under an average power
factor of 0.6 and PUE of 1.2. This corresponds to {carbon_mt:.3f} Mt CO2e. The estimate
describes the matched cohort under stated assumptions and does not represent global HPC emissions.
""".strip()
print(paragraph)
```
